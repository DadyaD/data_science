### Подготовка данных для обучения моделей

Разобьем выборку на обучающую, валидационную и тестовую

```python
target = data['average_fare']
features = data.drop(['average_fare'],axis=1)
features_train, features_valid, target_train, target_valid = train_test_split(
        features, target, test_size=0.40, random_state=3424)
features_test, features_valid, target_test, target_valid = train_test_split(
        features_valid, target_valid, test_size=0.50, random_state=3424)
```

Произведем масштабирование признаков

```python
scaler = StandardScaler()
features_train_scaled = features_train.copy()
features_valid_scaled = features_valid.copy()
features_test_scaled = features_test.copy()
scaler.fit(features_train)
features_train_scaled = scaler.transform(features_train)
features_valid_scaled = scaler.transform(features_valid)
features_test_scaled = scaler.transform(features_test)
pd.options.mode.chained_assignment = None
```

Модель дерева решений

```python
best_model_tree= None
best_result_tree = 10000
best_depth_tree = 0
best_leaf_tree = 0
best_r2_tree = 0
for depth in range(1, 20):
    for leaf in range(1,5):
        model_tree =DecisionTreeRegressor(random_state=3424,max_depth=depth, min_samples_leaf= leaf)
        model_tree.fit(features_train_scaled,target_train) 
        predictions_tree = model_tree.predict(features_valid_scaled)
        r2_tree = r2_score(target_valid, predictions_tree)
        result_tree =root_mean_squared_error(target_valid, predictions_tree)
        if result_tree < best_result_tree:
            best_model_tree = model_tree
            best_r2_tree = r2_tree
            best_result_tree = result_tree
            best_depth_tree = depth
            best_leaf_tree = leaf    
print('RMSE лучшей модели:',best_result_tree, 
      'глубина дерева',best_depth_tree,'минимальное количество выборок', best_leaf_tree, 'лучший коэфициент детерминации:',best_r2_tree)
```

Модель случайного леса

```python
best_model_forest = None
best_result_forest =100
best_est_forest = 0
best_depth_forest = 0
best_r2_forest=0
for est in range(50,70,10):
    for depth in range (1,5):
        model_forest = RandomForestRegressor(random_state=3424,n_estimators=est,max_depth=None)
        model_forest.fit(features_train_scaled, target_train)
        predictions_forest = model_forest.predict(features_valid_scaled)
        result_forest = root_mean_squared_error(target_valid, predictions_forest)
        r2_forest = r2_score(target_valid, predictions_forest)
        if result_forest < best_result_forest:
            best_model_forest = model_forest
            best_r2_forest=r2_forest
            best_result_forest = result_forest
            best_est_forest = est
            best_depth_forest = depth
print('RMSE лучшей модели:',best_result_forest,
      "Количество деревьев:", best_est_forest,'Глубина леса:',best_depth_forest, 'лучший коэфициент детерминации:',best_r2_forest)
```

Модель линейной регрессии

```python
model = LinearRegression(positive=True)
model.fit(features_train_scaled,target_train) 
predictions_lin_reg= model.predict(features_valid_scaled) 
result = root_mean_squared_error(target_valid,predictions_lin_reg)
r2_lin_reg = r2_score(target_valid, predictions_lin_reg)
print('RMSE:',result,'лучший коэфициент детерминации:',r2_lin_reg)
```
